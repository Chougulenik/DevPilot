```python
import os
from google import genai
from dotenv import load_dotenv

load_dotenv('.env')
api_key = os.getenv('GOOGLE_API_KEY')

# The Client automatically picks up your API key from the GEMINI_API_KEY environment variable.
# You can also pass it explicitly: client = genai.Client(api_key="YOUR_API_KEY")
client = genai.Client(api_key=api_key)

result = client.models.embed_content(
    model="text-embedding-004",
    contents="Text to embed"
)

# Extract the embedding vector
embedding = result.embeddings[0].values
print(f"Embedding length: {len(embedding)}")
```

```python
from google import genai

# The Client automatically picks up your GEMINI_API_KEY environment variable
client = genai.Client()

print("--- Available Embedding Models ---")

# client.models.list() returns a paginated iterator of Model objects
for model in client.models.list():
    # Filter to only show models that support embeddings
    if "embedding" in model.name.lower() or "embedContent" in getattr(model, 'supported_generation_methods', []):
        print(f"Model Name: {model.name}")
        
        # Optionally print descriptions or limits if they exist on the model object
        if hasattr(model, 'description'):
            print(f"Description: {model.description}")
        if hasattr(model, 'input_token_limit'):
            print(f"Max Input Tokens: {model.input_token_limit}")
            
        print("-" * 40)
```

```python
import sqlite3
conn = sqlite3.connect('db/chroma_db/chroma.sqlite3')
cur = conn.cursor()
cur.execute('SELECT id, name, topic FROM collections;')
print('Collections:', cur.fetchall())
cur.execute('SELECT id, collection_id FROM segments;')
print('Segments:', cur.fetchall())
cur.execute('SELECT segment_id, count(*) FROM embeddings GROUP BY segment_id;')
print('Embeddings per segment:', cur.fetchall())
```
